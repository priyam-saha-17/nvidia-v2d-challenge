# Reconstruction checkpoint access
Add a Kaggle Secret named `HF Access Key`, enter the read token for the approved Hugging
Face account, and enable access for this private notebook. Never paste the token
into a code cell. This notebook only checks checkpoint access; inference is not
implemented here yet.

```python
from kaggle_secrets import UserSecretsClient
from huggingface_hub import HfApi, hf_hub_url, get_hf_file_metadata
from pathlib import Path
import json
hf_token=UserSecretsClient().get_secret('HF Access Key')
api=HfApi(token=hf_token)
verified=[]
for repo in ['nvidia/cari4d_commercial','facebook/sam-3d-body-dinov3']:
 info=api.model_info(repo)
 weight=next(f.rfilename for f in info.siblings if f.rfilename.endswith(('.pt','.pth','.safetensors','.ckpt')))
 get_hf_file_metadata(hf_hub_url(repo,weight),token=hf_token)
 verified.append(repo)
Path('/kaggle/working/checkpoint_access.json').write_text(json.dumps({'verified':verified},indent=2))
print('Checkpoint access verified for:',', '.join(verified))
```
